```python
#Generate Random Coupon Numbers


coupons = np.random.randint(1, 6, 1000)
```

```python
#Count Coupon Occurrences

coupon_counts = pd.Series(coupons).value_counts().sort_index()
print(coupon_counts)
```

```python
#Visualize Uniform Distribution

plt.bar(coupon_counts.index, coupon_counts.values)
plt.title("Uniform Distribution of Coupons")
plt.xlabel("Coupon Number")
plt.ylabel("Frequency")
plt.show()
```

```python
#Calculate Probability of Each Coupon
probabilities = coupon_counts / coupon_counts.sum()
print(probabilities)
```

```python
#Display Uniform Distribution Summary

uniform_summary = pd.DataFrame({
    "Coupon": coupon_counts.index,
    "Probability": probabilities.values
})

uniform_summary
```

```python
uniform_summary = pd.DataFrame({
    "Coupon": coupon_counts.index,
    "Probability (%)": probabilities.values*100
})

uniform_summary
```

```python
#Now lets Select Review Count Column

review_count = retail["Review_Count"]
```

```python
#Create Review Count Ranges

review_groups = pd.cut(review_count, bins=10)
```

```python
#Count Records in Each Range

review_distribution = review_groups.value_counts().sort_index()
print(review_distribution)
```

```python
#Visualize Review Count Distribution

review_distribution.plot(kind='bar')
plt.title("Review Count Distribution")
plt.xlabel("Review Count Range")
plt.ylabel("Frequency")
plt.show()
```

```python
#Calculate Probability of Each Range

review_probability = review_distribution / review_distribution.sum()
print(review_probability)
```

```python
#Create Uniform Distribution Summary Table

uniform_summary = pd.DataFrame({
    "Review Count Range": review_distribution.index.astype(str),
    "Frequency": review_distribution.values,
    "Probability": review_probability.values
})

uniform_summary
```

#Task 2: Understanding Binomial Distribution


```python
#Import Visualization Library

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
```

```python
#Define Number of Customers

n = 20
```

```python
#Define Purchase Probability

p = 0.30
```

```python
#Display Trial Information

print("Number of Customers:", n)
print("Purchase Probability:", p)
```

```python
#Calculate Probability of Exactly 5 Purchases

prob_5 = binom.pmf(5, n, p)
print(prob_5)
```
