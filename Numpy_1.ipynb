#Reshaping of Array

```python
a = np.array([1,2,3,4,5,6])  # Coverting 1D array into 2D
print(a.reshape(3,2))
```

```python
# Converting into 3D
print(a.reshape(3,2))
```

```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
result = np.concatenate((arr1, arr2))
result
```

```python
# Concatenation of 2D Array
arr1 = np.array([[1,2],
[3,4]])
arr2 = np.array([[5,6],
[7,8]])

result1 = np.concatenate((arr1,arr2), axis=1)   #Horizontally JOins
result2 = np.concatenate((arr1,arr2), axis=0)   #Vertically JOins

print(result1)
print(result2)
```
